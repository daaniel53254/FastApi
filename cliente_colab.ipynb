# Cliente de la API de tareas
Consume la API (FastAPI en Docker) a través de ngrok. Pega abajo la URL pública del túnel.

```python
import requests

BASE_URL = "https://uninsured-untreated-skinning.ngrok-free.dev"  # URL de ngrok (sin barra final); cámbiala si el túnel cambia
HEADERS = {"ngrok-skip-browser-warning": "1"}
```

## POST /tareas: crear una tarea

```python
r = requests.post(f"{BASE_URL}/tareas", json={"titulo": "Probar desde Colab", "hecha": False}, headers=HEADERS)
print(r.status_code)
print(r.json())
id_creada = r.json()["id"]
```

## GET /tareas: listar todas

```python
r = requests.get(f"{BASE_URL}/tareas", headers=HEADERS)
print(r.status_code)
print(r.json())
```

## GET /tareas/{id}: consultar una tarea por id

```python
r = requests.get(f"{BASE_URL}/tareas/{id_creada}", headers=HEADERS)
print(r.status_code)
print(r.json())
```

## DELETE /tareas/{id}: borrar la tarea

```python
r = requests.delete(f"{BASE_URL}/tareas/{id_creada}", headers=HEADERS)
print(r.status_code)
print(r.json())
```
